# Task 1
## write a program to produce Fibonacci numbers up to 1000000

```python
# starting values
x1 = 0
x2 = 1
x3 = x1 + x2
print(x3)

while x3 <= 100000:
    # shift over values
    x1 = x2
    x2 = x3
    x3 = x2 + x1
    # display result
    print(x3)
```

# Task 2
## write a program to simulate 1000 tosses of a fair coin (use np.random.binomial)
## Calculate the mean and standard deviation of that sample

```python
import statistics

# lamda np.random.binomial on 1000 entries in an array, store them to calculate
# np.rand.bi takes (num of trials, 
coins = [0]*1000
coin_toss = list(map(lambda x: np.random.binomial(1, 0.5), coins))
print(coin_toss)

# calculate mean
print(statistics.mean(coin_toss))

# calculate std dv
print(statistics.stdev(coin_toss))
```

# Task 3
## Produce a scatterplot of y = 0.5*x+e where x has gaussian (0, 5) and e has gaussian (0, 1) distributions 
### use numpy.random.normal to generate gaussian distribution

```python
import matplotlib.pyplot as plt

# define x and y
# normal means gaussian
# ex: 
    # 1000 samples with a mean of 50 and standard deviation of 5
    # vector = rng.normal(loc=50, scale=5, size=1000)
x = np.random.normal(loc=0, scale=5, size=1000)
e = np.random.normal(loc=0, scale=1, size=1000)
y = 0.5 * x + e

# scatter plot
plt.scatter(x, y)
plt.show()
```
